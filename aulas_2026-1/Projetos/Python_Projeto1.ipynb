# O Cálculo de $\pi$ com o método da agulha de Buffon

Nesse primeiro projeto da disciplina de Introdução a Python - semestre 2026/1 - vamos fazer uma estimativa de $\pi$ com um método precursor do cálculo com simulação de Monte Carlo, com variáveis aleatórias. Esse método foi desenvolvido no século XVIII, por Georges-Louis Leclerc, Comte de Buffon.


## Formulação do problema

Suponha que temos um piso feito de tábuas paralelas de madeira, todas com a mesma largura, e deixamos cair uma agulha sobre o piso. Qual é a probabilidade de que a agulha fique atravessada sobre uma linha entre duas tábuas?

### A condição para que uma agulha cruze uma linha

- Se considerarmos o tamanho da agulha $l$, menor ou igual à distância entre as linhas paralelas, $d$,

$$l \leq d $$.

- Considerando o ângulo $\theta$ como sendo o ângulo agudo entre a agulha e a linha,
   - considerando a posição de uma das extremidades da agulha ao longo do eixo $x$, perpendicular à linha, **temos como condição para que a agulha cruze a linha**:



$$ \boxed{x \leq lsen(\theta)} $$

### A estimativa da probabilidade

A probabilidade de que uma dada agulha cruze uma das linhas será dada considerando-se a posição $x$ da agulha e o ângulo $\theta$, notando que as duas grandezas são independentes.

- Ao jogarmos uma agulha aleatoriamente, a probabilidade que ela pare em uma dada posição $x$, deve ser considerada como uma chance sobre todas as possibilidades de valores do eixo x, $d$ (distância entre as linhas). Logo a probabilidade de uma agulha parar em um determinado valor de $x$, é dada por $\frac{1}{d}$
  - A probabilidade, $f_X(x)$ da agulha se encontrar em uma dada posição é: $ f(x) = \frac{1}{d}$, para $0 \leq x \leq d$
- A probabilidade, $f_\Theta(\theta)$, da agulha estar em um determinado ângulo $\theta$ é: $f_\Theta(\theta) = \frac{1}{\pi}$, para $0 \leq \theta \leq \pi$

Considerando $x$ e $\theta$ como grandezas independentes (o valor de $x$ não interfere no de $y$ e vice-versa), a probabilidade total de uma agulha cruzar uma das linhas, será dada pela integral do produto de $f_X(x)$ e $f_\Theta(\theta)$ em todos os valores possíveis de $x$ e $\theta$, dentro das condições de cruzar a linha:

$$ P = \int_{0}^{\pi}\int_{0}^{lsen(\theta)} f_X(x) f_\Theta(\theta) dx d\theta = \int_{0}^{\pi}\int_{0}^{lsen(\theta)} \left( \frac{1}{d} \cdot \frac{1}{\pi} \right) dx d\theta $$

#### Integrando em $x$

$$\int_{0}^{lsen(\theta)} \left( \frac{1}{\pi} \cdot \frac{1}{d} \right) dx = \left[ \frac{x}{\pi d} \right]_0^{lsen\theta} = \frac{1}{\pi d} \left( lsen\theta - 0 \right) = \frac{lsen\theta}{\pi d}$$

#### Integrando em $\theta$

$$ P = \int_{0}^{\pi} \left( \frac{lsen\theta}{\pi d} \right) d\theta  = \frac{l}{\pi d} \int_{0}^{\pi} sen\theta d\theta$$

Como $\int_{0}^{\pi} sen\theta d\theta = 2$, temos que:

$$ \boxed{P = \frac{2l}{\pi d}} $$

Ou seja, a probabilidade de que uma agulha cruze uma das linhas é inversamente propocional a $d \cdot \pi$.

\
==> **Sendo assim, se tivermos uma forma de estimar a probabilidade de que uma agulha cruze a linha, teremos uma estimativa indireta do valor de $\pi$.**
