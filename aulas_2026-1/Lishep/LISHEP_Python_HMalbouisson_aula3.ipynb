## Calculando $\pi$ com o método (simplificado) de Arquimedes

O método de Arquimedes para determinação de $\pi$ consistiu em considerar um hexágono inscrito e outro circunscrito em um círculo.
* O perímetro do hexágono inscrito fornece uma aproximação inferior da estimativa;
* O perímetro do hexágono circunscrito fornece uma estimativa superior;
* Ao dobrar o número de lados do hexágono e obtendo a estimativa do perímetro dos polígonos de 12 lados inscritos e circunscritos, Arquimedes melhorou a estimativa de $\pi$;
* O fez assim, sucessivamente considerando um polígono de 12, 24, 48 e 96 lados;
* Conclui que $\pi$ seria um valor entre $\frac{223}{71}$ e $\frac{22}{7}$, ou seja, de acordo com Arquimedes, $\pi$ é um valor entre 3,1429 to 3,1408.

Vamos adotar um método simplificado para o cálculo de $\pi$ onde só consideramos o hexágono inscrito.

1. Considerando um hexágono inscrito em um círculo de raio R = 1
1. considerando que o hexágono é composto de 6 triângulos equiláteros
1. Considerando que 2$\pi$ R = 6*L, onde L é o lado do hexágono que é igual ao raio do círculo, L = R = 1 $\Rightarrow$ $$\pi = \frac{C}{D} = \frac{6}{2} = 3$$ $\Rightarrow$ aproximação de primeira ordem de $\pi$
1. Para melhorar a aproximação, podemos aumentar cada vez mais o número de lados do polígono inscrito, dobrando o número de lados por vez.
1. Tudo que devemos fazer é encontrar a expressão para o lado dos outros polígonos inscritos no círculo (12, 24, 48,... lados), utilizando o hexágono e trigonometria.

![](https://raw.githubusercontent.com/malbouis/Python_intro/master/hexagono_inscrito.png)

Considerando do hexágono inscrito no círculo, podemos deduzir o tamanho do lado de um polígono de 12 lados e fazer o mesmo para os outros polígonos, iterativamente.



$$ a = \sqrt{1 - \left(\frac{S_1}{2}\right)^2} \;\;\;\;\;\;\;\;\;\;\;\; (1)$$
$$ b = 1 - a \;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\; (2)$$
$$ S_{2} = \sqrt{b^2 + \left(\frac{S_1}{2}\right)^2} \;\;\;\;\;\;\;\;\; (3)$$
$$ \pi = \frac{C}{D} = \frac{N_{lados}*S_1}{2R} \;\;\;\;\;\;\;\;\;\;\;\;\; (4)$$



```python
import math

# hexágono
lado = 1.
novo_lado = 1.
n_lados = 6
pi = (n_lados*novo_lado)/2
print(pi)

#12 lados
n_lados = 12
a = math.sqrt(1 - pow(lado/2, 2))
b = 1 - a
novo_lado = math.sqrt(b**2 + pow(lado/2, 2))
pi = (n_lados*novo_lado)/2
print(pi)
```

Podemos incluir o cálculo acima em um loop e determinar pi para polígonos de N lados, começando com o hexágono.

```python
import math

R = 1.
lado = 1.
novo_lado = 1.
n_lados = 6.
for i in range(50):
  a = math.sqrt(1 - pow(lado/2, 2))
  b = 1 - a
  perimetro = n_lados*lado
  diametro = 2*R
  novo_lado = math.sqrt(b**2 + pow(lado/2, 2))
  pi = perimetro/diametro

  print(i, n_lados, lado, a, b, novo_lado, perimetro, pi)

  lado = novo_lado              #atualização do valor do lado do polígono
  n_lados = n_lados*2

print(math.pi)
```

Vamos encapsular o código que calcula $\pi$ e transformá-lo em uma função.

```python
import math

def pi_arq_simples(n=25, R=1):
  '''
  Essa função estima o valor de pi a partir de um hexágono inscrito em um círculo.
  É uma forma simplificada do método de Arquimedes do cálculo de pi.
  A função retorna duas listas, uma com o número de lados do polígono e a segunda com a estimativa de pi.
  A função tem como parâmetros:
  n: número total de polígonos usados para o cálculo
  r: raio do círculo (padrão é igual a 1)
  '''
  R = 1.
  lado = R
  novo_lado = R
  n_lados = 6.
  poligonos = []
  estimativa_pi = []
  for i in range(n):
    a = math.sqrt(1 - pow(lado/2, 2))
    b = 1 - a
    perimetro = n_lados*lado
    diametro = 2*R
    novo_lado = math.sqrt(b**2 + pow(lado/2, 2))
    pi = perimetro/diametro
    lado = novo_lado              #atualização do valor do lado do polígono
    n_lados = n_lados*2           # dobrar o numero de lados do poligono
    poligonos.append(n_lados/2)
    estimativa_pi.append(pi)

  print(f'Valor estimado de pi: {pi}. Valor tabelado: {math.pi}')

  return poligonos, estimativa_pi

#chamada da função
pi_arq_simples(25, 1.)
```

Por fim, podemos usar a função que criamos para ver como evolui a estimativa de pi com o método de Arquimedes.

```python
import matplotlib.pyplot as plt
import numpy as np
import math

pol, pi = pi_arq_simples(n=10, R=1.)

line = [math.pi for i in range(int(pol[-1]))]  # list comprehension

plt.plot(pol, pi)
plt.plot(line)
plt.xscale("log")
plt.xlabel("lados do polígono")
plt.ylabel("pi")
```

## Calculando $\pi$ com o Método de Monte Carlo

Nesse método, mostrado durante a [palestra](https://indico.cern.ch/event/869979/contributions/4239832/) do Prof. Luiz Mundim, sobre Métodos de Monte Carlo, é possível obter uma estimativa de $\pi$ baseada no uso de números aleatórios.

<center>
<img src=https://raw.githubusercontent.com/malbouis/Python_intro/master/circulo_inscrito.png width="200">
</center>

* Consideremos um círculo inscrito em um quadrado.
* A área do círculo, $A_C$, é dada por: $\pi r^2$, onde $r$ é o raio do círculo
* A área do quadrado, $A_Q$, é dada por: $(2*r)^2$
* Ao tomarmos a razão entre as áreas, temos que:
$$\frac{A_C}{A_Q} = \frac{\pi r^2}{(2*r)^2} = \frac{\pi}{4}$$

Também podemos estimar o valor da razão da área do círculo sobre o quadrado, $\frac{A_C}{A_Q}$, "jogando" um grande número de "bolinhas" no quadradro com o círculo inscrito e contando o número de vezes que as bolinhas caem dentro do círculo.

Vamos usar Python para fazer isso!

```python
import math
import random
from matplotlib import pyplot as plt

raio = 1.                                  #raio do círculo
A_c = 0                                    # inicializando o valor da área do círculo
total_bolinhas = 100000

# coordenadas das bolinhas. aleatórias.
for dots in range(0,total_bolinhas):
  x = random.uniform(-raio,raio)
  y = random.uniform(-raio,raio)
  # contando quantas bolinhas caem no círculo:
  distancia = math.sqrt(x**2 + y**2)            # distancia do centro do quadrado
  if (distancia < raio):
    A_c += 1

print(f'Pi = {4*(A_c/total_bolinhas)}')
```

Vamos agora incluir as bolinhas em um gráfico:


```python
import math
import random
from matplotlib import pyplot as plt         # https://matplotlib.org/stable/tutorials/introductory/pyplot.html

raio = 1.                                  #raio do círculo
A_c = 0                                      # inicializando o valor da área do círculo
total_bolinhas = 10000
x = []
y = []
x_c = []
y_c = []

# coordenadas das bolinhas. aleatórias.
for bolinha in range(0,total_bolinhas):
  x.append(random.uniform(-raio,raio))
  y.append(random.uniform(-raio,raio))
  # contando quantas bolinhas caem no círculo:
  distancia = math.sqrt(x[bolinha]**2 + y[bolinha]**2)            # distancia do centro do quadrado
  if (distancia < raio):
    A_c += 1
    x_c.append(x[bolinha])
    y_c.append(y[bolinha])

print(f'Pi = {4*(A_c/total_bolinhas)}')

plt.plot(x,y, 'o')
plt.plot(x_c,y_c, 'o')
```

Também podemos ver a distribuição da estimativa de $\pi$ para um certo número de bolinhas que jogamos.

```python
import math
import random
from matplotlib import pyplot as plt         # https://matplotlib.org/stable/tutorials/introductory/pyplot.html

def pi_mc(total_bolinhas):
  raio = 1.                                  #raio do círculo
  A_c = 0                                      # inicializando o valor da área do círculo
  x = []
  y = []
  x_c = []
  y_c = []

  # coordenadas das bolinhas. aleatórias.
  for bolinha in range(0,total_bolinhas):
    x.append(random.uniform(-raio,raio))
    y.append(random.uniform(-raio,raio))
    # contando quantas bolinhas caem no círculo:
    distancia = math.sqrt(x[bolinha]**2 + y[bolinha]**2)            # distancia do centro do quadrado
    if (distancia < raio):
      A_c += 1
      x_c.append(x[bolinha])
      y_c.append(y[bolinha])

  return 4*(A_c/total_bolinhas)

n_bolinhas = 10000
print(pi_mc(n_bolinhas))

pi_dist=[]
for i in range(1000):
  est_pi = pi_mc(n_bolinhas)
  pi_dist.append(est_pi)

plt.hist(pi_dist)
plt.xlabel("estimativa de pi")
```
