## Variáveis:

Uma varíavel é um nome que se refere a um valor. É uma forma poderosa de manipular valores em Python.

Consideremos a seguinte expressão matemática:

```
8*4
```

Podemos fazer uma **declaração de atribuição**. Isso consiste em **criar uma variável** que armazena o resultado da operação.

```
area = 8*4
print(area)
```

Ou seja, criamos uma variável que chamamos de `area`, à qual atribuímos o resultado da expressão `8*4`.

### Conceitos:

* **Expressão**: é uma combinação de valores, varíaveis e operadores;
* **Declaração**: é uma unidade de código que tem um efeito, como criar uma varíavel ou mostrar na tela um valor. Em geral, declarações não têm valores.

### Operações Matemáticas

Uma vez que as variáveis armazenam valores (de diferentes **tipos**), podemos usá-las para várias operações, como por exemplo, para realizar **Operações Matemáticas**.

Podemos calcular a área de um círculo:

```
import math

raio_circulo = 2.5
pi = math.pi
area_circulo = pi*raio_circulo*raio_circulo
area_circulo_v1 = pi*raio_circulo**2
area_circulo_v2 = pi*pow(raio_circulo, 2)
area_circulo_v3 = 3.14*2.5**2

print('area_circulo = ', area_circulo)
print('area_circulo_v1 = ', area_circulo_v1)
print('area_circulo_v2 = ', area_circulo_v2)
print('area_circulo_v3 = ', area_circulo_v3)
```

### Tipos de dados de Python: Booleano

É um outro **tipo** de dados em Python, usado como o ***resultado de uma proposição lógica***.

É muito usado em programação para tomada de decisões, como:

"Se essa expressão for verdadeira, faça uma tarefa. Se for falsa, faça outra tarefa."

A lógica acima representa uma **execução condicional** e aprenderemos como programá-las mais adiante.

Voltando aos booleanos, podemos também testar o tipo de uma variável booleana:

```
var = True
type(var)
```

```
res = 3 > 5   # aqui, o lado direito da declaração (3 > 5) é chamado de uma expressão booleana.
print(res)
```

### Operadores Lógicos

Operador |  resultado da expressão
--- | ---
**and** | Verdadeira, se ambas as comparações forem verdadeiras
**or** | Verdadeira, se pelo menos uma das expressões for verdadeira
**not**  | negativo da expressão

## Estruturas condicionais

Se quisermos que o nosso programa aja de forma diferente, de acordo com uma dada condição, precisaremos usar uma **lógica de condição**.

Por exemplo, para decidir se um dado número inteiro é par ou ímpar, podemos:
* Dividir o número por dois e
   * **se** o resto da divisão for igual a zero, é par;
   * **caso contrário**, é ímpar.

```
num_int = 9          #aqui, assumimos que o número seja mesmo inteiro, mas você pode adicionar uma proteção para somente pegar a parte inteira de qq número
resto_div = num_int % 2
print('resto da divisão:', resto_div)

if resto_div == 0:
  print('O número ', num_int, 'é PAR!')
else:
  print(f'O número {num_int} é IMPAR')
```

A estrutura condicional então é formada:
*   pela palavra reservada `if` seguida por uma condição booleana e por dois pontos (`:`)
*   as linhas abaixo dela formam o **bloco de instruções** que serão executadas se a condição for atendida. Para isso, elas devem ser **identadas corretamente, respeitando a especificação do Python**.

A **identação em Python** é feita com **exatamente 4 espaços**, para indicar as instruções que fazem parte do bloco de instruções caso a condição seja satisfeita.

##**Loops de repetição**
 Nos permite percorrer os itens de uma lista e, para cada um deles, executar um bloco de código.


```
nomes = ['Lishep', 'sessão', 'A', '2021']
for n in nomes:
    print(n)
```

A variável definida na linha 1 é uma lista inicializada com uma sequência de valores do tipo **string**.

A instrução `for` percorre todos esses elementos, um por vez e, em cada caso, atribui o valor do item à variável **n**, que é impressa em seguida.

O resultado, então, é a impressão de todos os nomes contidos na lista.

Isso continua até que o Python chegue ao último item da lista **nomes** -  '2021'. Depois disso, não há mais itens, portanto o `loop` é interrompido.

A palavra-chave `for` compõe a sintaxe usada para criar esse tipo de `loop`, por isso geralmente é chamada apenas de `loop for`.



##**Loops condicionais**

Os loops condicionais **se repetem até que algo aconteça**. Quando a condição testada passa a ser `False`, a execução é interrompida, saindo do `loop`, e passa para o próximo bloco.

No código a seguir, vemos um exemplo de uso do `loop while`, onde
* a condição testada é se a variável `contador` é menor que o valor 4
* enquanto essa expressão booleana não resultar em `True`, as instruções no `corpo do loop` serão executadas.

```
contador = 0
while contador < 4:
    print("O contador é:",contador)
    contador = contador + 1          # essa instrução é crucial em um loop while!!!! Aqui é onde o contador é atualizado!
```

* `contador = 0`
   * Antes de iniciarmos o `loop`, criamos uma variável chamada `contador` e atribuímos a ela um valor zero.
* `while contador < 4:`
   * A primeira atuação do `loop` é testar o valor atual da variável `contador`: é menor que 4?
   * Nesse ponto do loop, o valor da variável é 0, então a condição é `True`.

* Na linha seguinte, a variável `contador` recebe o incremento de 1 sucessivas vezes.
   * Chamamos isso de **atualização de valor**.
   * Quando o valor da variável `contador` é 4 e o `loop while` não é mais verdadeiro, o loop é interrompido.

* Caso a condição de parada nunca seja
atingida, o loop será executado infinitamente, gerando problemas no programa.

* Estruturas de controle, condicionais e de repetição, estão presentes na maioria das linguagens de programação e representam uma parte fundamental de cada uma delas. Sendo assim, é muito importante conhecer a sintaxe e o funcionamento dessas estruturas.



* As instruções `break` e `continue` são ferramentas das estruturas de repetição que permitem a interrupção de um ciclo do `loop` ou do `loop` como um todo.
   * A instrução `break` interrompe todo o `loop`,
   * A instrução `continue` finaliza uma única iteração, fazendo com que a execução volte para o início do bloco do `loop`.

No exemplo a seguir, vamos determinar quais são os números pares até o número 10.




```
i = 0
while i <= 10:
    i+=1               # i = i + 1
    if i % 2 != 0:     # apenas números pares
        continue
    print(i)
```

**Vimos até aqui**:
* **Estruturas condicionais** (`if`, `else`): uma estrutura de condição que permite avaliar uma expressão e, de acordo com seu resultado, executar uma determinada ação.
* `loops`: permite executar um bloco de código repetidas vezes. Também é chamado de `laço` em português
 * **condicionais**: continua executando enquanto não houver uma condição determinada (`while`);
 * **repetição**: repete uma instrução, um certo número de vezes (`loop`).


## **Funções** em Python

Suponha que temos uma série de instruções que repetimos em diferentes lugares do nosso programa.

Por exemplo, já testamos se um número é par duas vezes nessa aula.

Podemos **encapsular** esses comandos em uma **função**. Vejamos a sintaxe:

```
def par():
  num_int = 5
  if num_int % 2 != 0:
    print(f'o número {num_int} é IMPAR')
  else:
    print(f'o número {num_int} é PAR')
```

* `def par():` sintaxe de definição da função. O sinal `:` indica que as instruções da função vêm no **bloco de instruções** a seguir
* O corpo da função é identado.
   * Isso significa que todas as instruções seguintes ao sinal "`:`" que estiverem a 4 intervalos do início da linha, serão executadas.


Mas ainda não conseguimos ver o retorno da função. Até agora, tudo o que fizemos foi definir a função e suas instruções. Para ver o resultado, precisamos fazer a **chamada da função**.

```
par()  # chamada da função que criamos acima.
```

Mas da forma como construímos essa função, só testamos o número 5 (`num_int = 5`). Isso não é muito prático.

$\Rightarrow$ Vamos redefinir a nossa função de forma a incluir o número como **parâmetro** da função!

```
def par(num_int):
  if num_int % 2:
    print(f'o número {num_int} é IMPAR')
  else:
    print(f'o número {num_int} é PAR')

#chamadas da função
par(8)
par(9)
par(5)
```

No caso de querermos reutilizar o resultado da função em um outro cálculo, precisamos reescrever a nossa função para que ela **retorne um valor**.

Isso é feito com a instrução `return` dentro do bloco da função.

```
def par(num_int):
  num_par = True
  if num_int % 2 != 0:
    num_par = False
  return num_par

num = 6
print(par(num))
```

Para conseguir utilizar o valor de retorno da função em um outro cálculo, podemos atribuir o valor a uma variável:

```
par_result = par(num)
print(par_result)

if par_result:
  print(f'{num} é PAR')
else:
  print(f'{num} é IMPAR')
```

### Em resumo:
* Funções são definidas usando a sintaxe `def`;
* Funções são chamadas usando parênteses;
* Funções tomam parâmetros e podem retornar valores;
* Para reutilizar um valor de retorno de uma função, podemos salvá-lo em uma variável.
