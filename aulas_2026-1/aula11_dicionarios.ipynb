# Dicionários

Todos os tipos de **dados compostos** (***compounds***) que estudamos em detalhes até agora - strings, listas - são **tipos de sequência**, ou seja coleções ordenadas que usam inteiros como índices para acessar os valores que contêm dentro deles.

Os **dicionários** são outro tipo de dado composto. Eles são o **tipo de mapeamento** integrado do Python. Eles mapeiam **chaves** (**keys**), que podem ser de qualquer **tipo imutável**, para valores, que podem ser de qualquer tipo (heterogêneos), assim como os elementos de uma lista ou tupla.


**dicionários: chave -> valor**

```python
tradutor = {}                # atribuição de criação de um dicionário vazio
tradutor["one"] = "um"       # atribuição do par key:value
tradutor["two"] = "dois"
tradutor["three"] = "tres"
print(tradutor)
```

Os pares de `chave:valor` do dicionário são separados por vírgulas. Cada par contém uma chave e um valor separados por dois pontos (`:`).

```python
frutas = {"maçãs": 430, "bananas": 312, "laranjas": 525, "peras": 217, "goiabas": 88, "figos": 108, "jambos": 249}
print (frutas)
```

O construtor `dict` também constrói dicionários a partir dos pares `chave:valor` diretamente.

```python
tradutor_v2 = dict([("one", "um"), ("two", "dois"), ("three", "tres")])
print(tradutor_v2)
```

Quando as `keys`de um dicionário são `strings` simples:

```python
tradutor_v3 = dict(one="um", two="dois", three="tres")
print(tradutor_v3)
```

## Dicionários não são ordenados!!!! (a partir de Python versão 3.7, são ordenados)

Até Python v3.6, os itens do dicionário não eram necessariamente impressos na ordem de entrada.



```python
dicionario = {}
for i in range(200):
    dicionario[i] = "valor" + str(i)
print (dicionario)
```

```python
print(dicionario[48])
```

Mas apesar dos itens dos dicionários agora serem ordenados, a ordem dos ítens não determina inteiramente o objeto dicionário. Para dois objetos de tipo dicionário serem considerados iguais, basta terem os mesmos pares `chave:valor`.

```python
tradutor_v2 = dict([("one", "um"), ("two", "dois"), ("three", "tres")])
print(tradutor_v2)

tradutor_v3 = dict(one="um", three="tres", two="dois")
print(tradutor_v3)
```

```python
tradutor_v2 == tradutor_v3
```

Os valores ou keys não podem ser acessados via índices de posição.

```python
print(tradutor[1])
```

```python
print(tradutor_v2)
print(tradutor_v2["three"])
```

```python
pessoa = dict([("nome", "Ana"), ("idade", 28), ("cidade", "São Paulo"), ("ativo", True)])
print(pessoa)
```

Até agora, os tipos de dados que tínhamos visto, eram sequências, como listas e strings. Porém **os dicionários não são sequências**, sendo assim **não podemos indexá-los**. Dicionários são uma espécie de mapeamento.

### Compreensão de dicionário (dictionnary comprehension)

Também é possível se criar um dicionário utilizando compreensão de dicionário, como fazemos com listas.

```python
comp_dict = {x: x**2 for x in (2, 4, 6)}
print(comp_dict)
```

```python
print(comp_dict[6])
```

## Operações com dicionários

A instrução `del` retira um par `chave:valor` de um dicionário. Por exemplo, o dicionário a seguir contém os nomes de várias frutas e o número de cada fruta em estoque:

```python
frutas = {"maçãs": 430, "bananas": 312, "laranjas": 525, "peras": 217, "goiabas": 88, "figos": 108, "jambos": 249}
print(frutas)
```

```python
del frutas["peras"]
print(frutas)
```

```python
frutas["peras"] = 0
print(frutas)
```

```python
len(frutas)  #a função len também funciona com dicionários
```

## Métodos de dicionários

Dicionários têm vários métodos úteis.

Podemos visualizar as `keys` de um dado dicionário:

```python
frutas.keys()
```

```python
for fruta in frutas.keys():
    print(fruta, ' = ', frutas[fruta])
```

```python
print(frutas)
```

```python
lista_frutas = list(frutas.keys())
print(lista_frutas)
```

É tão comum iterar sobre as `keys` de um dicionário que podemos omitir a chamada do método de `keys` no loop `for`
   * **iterar sobre um dicionário itera implicitamente sobre suas chaves:**

```python
for fruta in frutas:
    print(fruta, frutas[fruta])
```

```python
keys_frutas = [fruta for fruta in frutas] # list comprehension
print(keys_frutas)
```

```python
valores_frutas = [frutas[fruta] for fruta in frutas]
print(valores_frutas)
```

O método **values** nos possibilita acessar diretamente os valores do dicionário:

```python
valores = (frutas.values())
print(valores)
```

O método `items` retorna uma tupla para cada par `(key:valor)`:

```python
print(frutas.items())
```

As tuplas podem ser úteis para obter a `key` e o valor ao mesmo tempo durante a execução do loop, ao utilizar o método `items`:

```python
for (k,v) in frutas.items():
    print("Encontrei a chave ",k," que mapeia para o valor ",v)
```

Os operadores `in` e `not in` podem testar se uma chave está no dicionário:

```python
"bananas" in frutas
```

```python
 "jambos" in frutas
```

```python
"mangas" in frutas
```

Esse método pode ser bastante útil pois ao tentar acessar uma `key` não existente, Python retorna um erro de execução:

```python
frutas["mangas"]
```

O método `get` é também bem útil. Ele retorna o valor da `key` especificada:

```python
matrix = {(0, 3): 1, (2, 1): 2, (4, 3): 3}
matrix.get((4,3))
```

Caso a `key` não exista, retorna `None`,  ou um valor indicado como segundo parâmetro:

```python
a = matrix.get((2,2))
print(a)
```

```python
b=matrix.get((2,2), 0)
print(b)

c=matrix.get((2,1),-999)
print(c)
```

```python
frutas.get("mangas","Não tem manga")
```

## Aliasing e cópia

Como no caso das listas, porque os **dicionários são mutáveis**, precisamos estar cientes do **aliasing** ou referência ao mesmo objeto.

Sempre que duas variáveis se referem ao mesmo objeto, as alterações em uma delas afetam a outra.

Se quiser modificar um dicionário e manter uma cópia do original, use o método de **cópia**  próprio do objeto, `copy`.

Consideremos um dicionário chamado **opostos**, que contém pares de opostos:

```python
opostos = dict(up="down", right="wrong", yes="no")    # criação de um dicionário
alias = opostos                                            # criação de um alias. o dicionário alias e opostos se referem ao mesmo objeto.
copia = opostos.copy()                                     # cópia do dicionario opostos
```

* Os dicionários ***alias*** e ***opostos*** se referem ao **mesmo objeto**;
* O dicionário ***copia*** se refere a uma nova cópia do mesmo dicionário.
* Se modificarmos o ***alias***, o dicionário ***opostos*** também será alterado:

```python
print(opostos)
print(alias)
```

```python
alias["right"] = "left"
print(opostos)
```

```python
print(copia)
```

```python
copia["goodbye"] = "hello"
print(copia)
```

```python
print(opostos)
```
